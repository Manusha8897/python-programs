```python
class Parent:
    def func1(self):
        print("this function is in parent class.")

    
class Child(Parent):
    def func2(self):
        print("This function is in Child class.")
object=Child()
object.func1()
object.func2()
```

```python
class Parent:
    def funC1(self):
        print("this function belongs to parent class")
class Child(Parent):
    def funC1(self):
        print("this function belongs to child class")
object=Child()
object.funC1()
```
